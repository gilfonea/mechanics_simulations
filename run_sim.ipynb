```python
import threading
from models.two_bodies_on_incline import Two_bodies_on_incline

# 1. אתחול המערכת (בונה את הקנבס והמשתנים, אבל לא נתקע בלולאה)
sim = Two_bodies_on_incline()

# 2. יצירת תהליכון (Thread) שיריץ את הלולאה האינסופית ברקע
animation_thread = threading.Thread(target=sim.start)

# 3. הפעלת התהליכון (הקוד ממשיך הלאה והמחברת מסיימת לרוץ)
animation_thread.start()
```
